```python
# 넘파이 배열
import numpy as np

aa= np.array([10,20,30])
bb= np.array([4,5,6])

# 배열끼리 + 연산 -- 백터값의 산술연산 됨 -- 같은 자리의 요소들끼리 덧셈
cc= aa+bb
print(cc)
```

```python
# 인덱싱과 슬라이싱
scores= np.array([5,4,6,2,8,0,3,4,7,8,1,2,9])

#1번 학생
first_student= scores[0]
#3~7번 쯤 중간쯤 학생
middle_student= scores[2:7]

print(first_student)
print(middle_student)

# 불리언 인덱싱
high_scores= scores[scores>=8]
print(high_scores)
print(high_scores.mean())
```

```python
sales_df= pd.DataFrame({
    'date': ['2026-01-01']*3 + ['2026-01-02']*3 + ['2026-01-03']*3,
    'region': ['s','b','d']*3,
    'sales': [120,90,70,80,20,30,502,125,64],
    'units': [12,50,7,14,10,8,1,11,10],
    'ad_cost': [30,40,20,65,40,30,30,50,np.nan]
})
sales_df.info()
```

```python
#결측치 확인
sales_df.isna()
```

```python
#결측치 수량 확인
sales_df.isna().sum()
```

```python
sales_df['ad_cost']= sales_df['ad_cost'].fillna(sales_df['ad_cost'].median())
sales_df.isna().sum()
```

```python
# 판매 지역별 평균 판매액
df= sales_df.groupby('region')['sales'].agg(['sum','mean'])
df
```

```python
# 시각화
import matplotlib.pyplot as plt
#선 그래프
plt.figure(figsize=(7,3))
plt.plot(df.index, df.values, marker='o')
plt.show()
```

```python
# 산점도 그래프
plt.figure(figsize=(7,3))
plt.scatter(sales_df['ad_cost'], sales_df['sales'], marker='o')
plt.grid(True)
plt.show()
```

```python
# 1. CSV 저장
sales_df.to_csv('aaa.csv', index=False, encoding='utf-8')

# 2. TSV(메모장 탭 구분) 저장
sales_df.to_csv('aaa.tsv', sep="\t", index=False, encoding='utf-8')

# 3. 엑셀 저장 (to_excal ➡️ to_excel로 수정)
sales_df.to_excel('aaa.xlsx', index=False)
```
